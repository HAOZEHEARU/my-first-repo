```python
text='''桃之夭夭，灼灼其华。
之子于归，宜其室家。
桃之夭夭，有蕢其实。
之子于归，宜其家室。
桃之夭夭，其叶蓁蓁。
之子于归，宜其家人。
'''
count = {}
for ch in text:
    if ch in count:
        count[ch]=count[ch]+1
    else:
        count[ch]=1
print(len(count))
print(count)
result = sorted(count.items(), key=lambda x: x[1], reverse=True)
for ch, n in result:
    print(f"{ch}:{n}", end=" ")
```
